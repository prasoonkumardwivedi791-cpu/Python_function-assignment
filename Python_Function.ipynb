1. What is the difference between a function and a method in Python?

   -> Function: A standalone block of code that performs a task and can be called independently. Defined with def and not tied to any object unless you place it inside a class.

   Method: A function defined inside a class and associated with an object. It operates on the data (attributes) of the object. Methods implicitly take the object reference (self) as their first parameter.

2. Explain the concept of function arguments and parameters in Python.

  -> Parameters: Variables listed in the function definition. They act as placeholders for the values the function will receive.

   Arguments: Actual values passed to a function when it is called.

5. What are iterators in Python and how do they differ from iterables?


   ->Iterator: An object that returns one item at a time when you call next() on it. It remembers its state during iteration.

   Iterable: Any object you can loop over. It does not produce values by itself — it must provide an iterator using __iter__().

   Core difference:
	 •	Iterable gives you an iterator.
	 •	Iterator gives you the next value.


6.Explain the concept of generators in Python and how they are defined.

   ->Generators produce values on-demand instead of holding everything in memory. They are used for efficient iteration, especially with large datasets or streams of data.

   A generator is basically a function that uses yield instead of return.

   Key traits
	 •	Lazy evaluation: values are produced one at a time
	 •	Memory-efficient: no need to store full lists
	 •	Maintain state between yields


7. What are the advantages of using generators over regular functions?

  ->Generators win when efficiency matters. They don’t build entire data structures in memory — they stream results one by one. That gives you power regular functions can’t match in large-scale tasks.

  Advantages


  1.Memory efficiency
	 •	They yield values one at a time instead of creating full lists.


  2.Better performance for large data
	 •	No upfront computation; only compute what you need when you need it.


	3.Lazy evaluation
	 •	They stop as soon as you’re done consuming values.
	 •	Ideal for pipelines and real-time processing.


	4.State persistence
	 •	They pause and resume execution automatically, remembering variables      between yields.


	5.Cleaner code for iterators
	 •	Easier than writing classes with __iter__() and __next__().



8. What is a lambda function in Python and when is it typically used?   

  ->A lambda function is a small, anonymous function defined with the lambda keyword instead of def. It can take any number of arguments but has only one expression

  	•	When you need a short, throwaway function
	  •	As a callback in higher-order functions
	  •	Inline with tools like map(), filter(), and sort()


9. Explain the purpose and usage of the `map()` function in Python.

  ->map() applies a function to each item in an iterable and returns a map object (an iterator). It lets you transform data without writing explicit loops.
  

  Purpose
	 •	Process each element in a sequence
	 •	Clean, functional-style data transformation
	 •	Often paired with lambda, but can use normal functions too
